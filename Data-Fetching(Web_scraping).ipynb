```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

num = 1
all_quotes_list = []

while True:
    url = f'http://quotes.toscrape.com/page/{num}/'
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36'}

    response = requests.get(url, headers=headers, timeout=10)
    soup = BeautifulSoup(response.text, 'lxml')

    quotes = soup.find_all('div', class_='quote')

    if not quotes:
        print("All pages are done!")
        break

    for q in quotes:
        text = q.find('span', class_='text').text
        author = q.find('small', class_='author').text
        tags = [t.text for t in q.find_all('a', class_='tag')]

        all_quotes_list.append({
            'quote': text,
            'author': author,
            'tags': ", ".join(tags)
        })

    print(f"Page {num} scraped...")
    num += 1  # to go on next page

# Final Step: Save to CSV
df = pd.DataFrame(all_quotes_list)
df.to_csv('final_quotes.csv', index=False)
print(f"Success! Total {len(df)} quotes saved.")
```
