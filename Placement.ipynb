```python
import numpy as np
import pandas as pd
path='/content/drive/MyDrive/ML_Datasets/student_placement_synthetic.csv'
df= pd.read_csv(path)
df.head()
```

```python
df.info()
```

```python
# Final Selection based on your logic
X = df[['cgpa', 'backlogs', 'coding_skills', 'dsa_score', 'aptitude_score',
        'communication_skills', 'system_design', 'internships',
        'projects_count', 'hackathons', 'open_source_contributions']]

y = df['placement_status']

# Preprocessing: Video wala step (Scaling)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Split karna (20,000 students ka test hoga)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 2. Model ko Training par lagana
model = LogisticRegression()
model.fit(X_train, y_train)

# 3. Model se predictions maangna (Test data par)
y_pred = model.predict(X_test)

# 4. Accuracy check karna
print("Model Accuracy:", accuracy_score(y_test, y_pred))
```

```python
my_stats = [[8.5, 0, 80, 85, 90, 80, 75, 2, 3, 0, 1]] # CGPA, backlogs, etc.
my_stats_scaled = scaler.transform(my_stats) # Scaling zaroori hai!

prediction = model.predict(my_stats_scaled)
print("Placed" if prediction[0] == 1 else "Not Placed")
```
