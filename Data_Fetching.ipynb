```python
import time
import requests
import pandas as pd

all_movies = []

for i in range(1, 429):
    url = f'https://api.themoviedb.org/3/movie/popular?api_key="YOUR API KEY HERE"&page={i}'
    response = requests.get(url)

    # 1. Check if the request was actually successful
    if response.status_code == 200:
        data = response.json()

        # 2. Safety check: does 'results' exist?
        if 'results' in data:
            temp_df = pd.DataFrame(data['results'])[['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']]
            all_movies.append(temp_df)
        else:
            print(f"Skipping page {i}: No 'results' found.")
    else:
        print(f"Error on page {i}: Status Code {response.status_code}")

    # 3. Rate limiting protection: pause every 50 requests
    if i % 50 == 0:
        time.sleep(1)

# Combine everything
df = pd.concat(all_movies, ignore_index=True)
df.head()
```
